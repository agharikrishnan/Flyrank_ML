## 2. The question: decision, action, cost of a wrong call

*Decision: Which content pages should a reviewer prioritize first for further review?

Who acts: A content or SEO reviewer uses the ranked output to decide which pages to inspect and whether to consider actions such as refreshing, expanding, protecting, pruning, or monitoring them.

Cost of a wrong call: Prioritizing a low-value page can waste limited reviewer time. Failing to surface a genuinely useful opportunity can mean that an important page is not reviewed when it could have benefited from attention.

Why data or ML could help: The dataset contains multiple observable signals about visibility, traffic, freshness, search position, and engagement. A simple rule may be sufficient, so I will compare against a baseline before assuming ML is necessary. If the signals interact in ways that are difficult to express with a fixed rule, ranking/scoring may provide useful decision support.*

```python
# Confirm the columns needed for the framing are present.
required = [
    "content_id",
    "client_id",
    "impressions_90d",
    "sessions_90d",
    "content_age_days",
    "avg_position",
    "trend_direction",
]

missing = [c for c in required if c not in df.columns]
print("Missing required columns:", missing)
```

## 3. Quick look at the data (2-3 real numbers)

*The starter dataset contains 30,000 content items. More than half currently have a downward trend, while the typical page has much less traffic than the highest-volume pages. This makes prioritization worth investigating: a reviewer cannot give the same level of attention to every page, so a ranked queue could help focus limited review capacity.*

```python
n_rows = len(df)
down_count = (df["trend_direction"] == "down").sum()
down_pct = down_count / n_rows * 100
median_impressions = df["impressions_90d"].median()
median_sessions = df["sessions_90d"].median()
print(f"Content items: {n_rows:,}")
print(f"Downward-trend items: {down_count:,} ({down_pct:.1f}%)")
print(f"Median 90-day impressions: {median_impressions:,.0f}")
print(f"Median 90-day sessions: {median_sessions:,.0f}")
```
