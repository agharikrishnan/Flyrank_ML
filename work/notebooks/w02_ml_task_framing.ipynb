## 1. My lane as an ML task (type)

*Task type: Classification*

*I chose the Refresh / Content Opportunity Scoring lane. I see this as a classification problem because I want to separate pages that are showing a decline from pages that are not. The main reason for doing this is to help decide which pages should be looked at first. I think classification makes sense here because the starter data already gives us a trend direction that can be turned into a yes/no target.*

```python
import pandas as pd
df = pd.read_csv("/content/Flyrank_ML/data/raw/content_refresh_anonymized.csv")
print(f"Rows: {len(df):,}")
print(f"Columns: {len(df.columns):,}")
```

## 2. Target or proxy

*For the target, I would use is_declining_label. I would make it 1 when trend_direction is "down" and 0 otherwise. I am treating this as a proxy because it describes what is happening in the current data, rather than predicting a future result. A better target later could use the current period to predict what happens in the next period.*

```python
df["is_declining_label"] = (
    df["trend_direction"] == "down"
).astype(int)

print(df["is_declining_label"].value_counts())
```

## 3. Success metric

*Success metric: Precision@50*

*I would use Precision@50 as the main metric. The reason is that the output is supposed to help make a review list. If someone only has time to look at 50 pages, I care about how many of those 50 pages are actually useful candidates. This seems more useful for this problem than just looking at overall accuracy.*

```python
target_rate = df["is_declining_label"].mean()
print(f"Overall declining proxy rate: {target_rate:.3f}")
print("Success metric: Precision@50")
```

## 4. The unit of analysis, as a real dataframe

*One row represents one content item. The content_id is the ID for that item. The other columns describe things that have been observed about the page, such as impressions, sessions, age and trend. This is the unit I care about because the final result should tell a reviewer which individual pages to look at first.*

```python

lane_df = df[
    [
        "content_id",
        "client_id",
        "impressions_90d",
        "sessions_90d",
        "content_age_days",
        "trend_direction",
        "is_declining_label",
    ]
].copy()
lane_df.head(10)
```

## 5. Why ML beats a fixed rule here

*A simple rule could just look for pages where the trend is down. That would be easy, but it would not look at the other things happening with the page. For example, a page might have a lot of impressions but low CTR, or it might be old but still doing well. I think ML could be useful because it can look at several signals together instead of relying on one condition. I would still compare it with a simple rule first, because if the rule works just as well then there is not much reason to use a more complicated model.*

```python
baseline_count = df["is_declining_label"].sum()
baseline_rate = df["is_declining_label"].mean()
print("Pages flagged by the simple decline rule:", baseline_count)
print(f"Share of pages flagged: {baseline_rate:.3f}")
```
